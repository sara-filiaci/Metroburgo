# Analisi del questionario sul progetto Metroborgo

## Obiettivo dell’analisi

Questo notebook analizza i risultati del questionario dedicato al progetto Metroborgo e alle percezioni relative a Montalto delle Marche.

L’obiettivo è descrivere separatamente:

* i residenti e frequentatori abituali del territorio;
* i visitatori non residenti o non abituali.

L’analisi prende in considerazione il profilo dei rispondenti, le caratteristiche dell’esperienza di visita, le percezioni sul territorio, gli effetti attribuiti al progetto e alcuni aspetti economici.

## Struttura del lavoro

Il procedimento comprende:

1. caricamento e controllo del dataset;
2. separazione dei residenti e dei visitatori;
3. rimozione delle colonne non pertinenti e dei questionari privi di risposte utili;
4. controllo delle compilazioni parziali;
5. analisi descrittiva delle risposte;
6. calcolo di frequenze, percentuali, medie e mediane;
7. raggruppamento delle affermazioni in aree tematiche.

Le percentuali sono calcolate sul numero di risposte valide disponibili per ciascuna domanda. Le risposte mancanti non vengono sostituite con valori stimati.

## Organizzazione del notebook

Il notebook è suddiviso in due sezioni principali:

* analisi dei visitatori;
* analisi dei residenti e frequentatori abituali.

Le interpretazioni conclusive e le visualizzazioni grafiche vengono sviluppate separatamente nella presentazione finale.

```python
# Importo la libreria Pandas
import pandas as pd

# Carico il file CSV
df = pd.read_csv("results-survey-07.csv")

# Visualizzo le prime cinque righe del dataset
df.head()
```

## Suddivisione del campione

Il questionario distingue gli intervistati in base alla domanda:

**“Vivi a Montalto delle Marche o la frequenti abitualmente?”**

Le risposte vengono utilizzate per creare due gruppi:

- **Residenti o frequentatori abituali**: risposte “Sì”
- **Non residenti o frequentatori non abituali**: risposte “No”

La suddivisione mantiene tutte le risposte associate a ciascun partecipante, poiché ogni riga del dataset rappresenta un questionario completo.

```python
# Nome della colonna utilizzata per suddividere il campione
colonna_residenza = (
    "Vivi a Montalto delle Marche o la frequenti abitualmente?"
)

# Controllo delle risposte presenti nella colonna
df[colonna_residenza].value_counts(dropna=False)
```

## Rimozione delle risposte mancanti e creazione dei gruppi

I due questionari privi di risposta alla domanda di suddivisione vengono esclusi dall’analisi comparativa.

Il campione valido viene quindi diviso in due DataFrame:

* residenti o frequentatori abituali;
* non residenti o frequentatori non abituali.


```python
# Nome della colonna utilizzata per suddividere il campione
colonna_residenza = (
    "Vivi a Montalto delle Marche o la frequenti abitualmente?"
)

# Creo una copia del dataset eliminando solamente le righe
# prive di risposta alla domanda di suddivisione
df_valido = df.dropna(subset=[colonna_residenza]).copy()

# Creo il DataFrame dei residenti o frequentatori abituali
df_residenti = df_valido[
    df_valido[colonna_residenza] == "Sì"
].copy()

# Creo il DataFrame dei non residenti o frequentatori non abituali
df_turisti = df_valido[
    df_valido[colonna_residenza] == "No"
].copy()

# Reimposto gli indici dei due DataFrame
df_residenti.reset_index(drop=True, inplace=True)
df_turisti.reset_index(drop=True, inplace=True)

# Verifico il numero di risposte nei due gruppi
print("Risposte valide totali:", len(df_valido))
print("Residenti o frequentatori abituali:", len(df_residenti))
print("Non residenti o frequentatori non abituali:", len(df_turisti))
```

## Controllo e pulizia iniziale del DataFrame dei turisti

Prima di analizzare le risposte, viene controllata la struttura del DataFrame dedicato ai turisti.

In particolare:

1. vengono visualizzate tutte le colonne;
2. viene contato il numero di risposte presenti in ciascuna colonna;
3. vengono individuate ed eliminate le colonne completamente vuote;
4. vengono cercate eventuali righe prive di risposte utili.

Il DataFrame originale `df_turisti` viene conservato, mentre le operazioni di pulizia vengono applicate a una copia.


```python
# Stampo tutti gli header presenti nel DataFrame dei turisti
for numero, colonna in enumerate(df_turisti.columns, start=1):
    print(f"{numero}. {colonna}")

print("\nNumero totale di colonne:", len(df_turisti.columns))
```

```python
# Conto il numero di valori non mancanti per ogni colonna
conteggio_risposte_turisti = df_turisti.notna().sum()

# Trasformo il risultato in una tabella più leggibile
riepilogo_colonne_turisti = (
    conteggio_risposte_turisti
    .reset_index()
)

riepilogo_colonne_turisti.columns = [
    "Colonna",
    "Numero di risposte"
]

# Aggiungo il numero di valori mancanti
riepilogo_colonne_turisti["Valori mancanti"] = (
    len(df_turisti)
    - riepilogo_colonne_turisti["Numero di risposte"]
)

riepilogo_colonne_turisti
```

### Rimozione delle colonne non pertinenti al gruppo dei turisti

Nel DataFrame dei turisti sono presenti alcune domande rivolte esclusivamente ai residenti. Poiché queste colonne non contengono alcuna risposta nel gruppo analizzato, vengono rimosse.

In questa fase non vengono eliminate righe, perché alcuni questionari potrebbero essere stati compilati solo parzialmente.


```python
# Individuo le colonne completamente vuote nel gruppo dei turisti
colonne_vuote_turisti = df_turisti.columns[
    df_turisti.isna().all()
].tolist()

# Creo una copia eliminando solamente le colonne completamente vuote
df_turisti_pulito = df_turisti.drop(
    columns=colonne_vuote_turisti
).copy()

print("Colonne originali:", df_turisti.shape[1])
print("Colonne completamente vuote eliminate:", len(colonne_vuote_turisti))
print("Colonne rimaste:", df_turisti_pulito.shape[1])
```

### Controllo delle colonne rimaste

Dopo aver eliminato le colonne completamente vuote, vengono esaminate le 51 colonne rimaste, visualizzandone il nome completo e il numero di risposte presenti.

Questo controllo permette di distinguere le informazioni tecniche, le domande rivolte ai turisti, i dati sociodemografici e le eventuali colonne anomale.


```python
# Mostro interamente il testo contenuto nelle colonne della tabella
pd.set_option("display.max_colwidth", None)
pd.set_option("display.max_rows", None)

# Creo il riepilogo delle colonne rimaste
riepilogo_turisti_pulito = pd.DataFrame({
    "Colonna": df_turisti_pulito.columns,
    "Numero di risposte": df_turisti_pulito.notna().sum().values,
    "Valori mancanti": df_turisti_pulito.isna().sum().values
})

riepilogo_turisti_pulito
```

### Rimozione delle colonne non pertinenti e tecniche

Dal DataFrame dei turisti vengono eliminate:

* le cinque colonne relative al cambiamento del valore patrimoniale, rivolte al gruppo dei residenti;
* le variabili tecniche generate dal sistema di raccolta;
* la domanda utilizzata per suddividere il campione, poiché tutti i partecipanti presenti in questo DataFrame appartengono già al gruppo dei non residenti.

L’identificativo della risposta viene conservato per permettere eventuali controlli sui singoli questionari.


```python
# Colonne tecniche che non servono per l'analisi
colonne_tecniche_turisti = [
    "Data invio",
    "Ultima pagina",
    "Lingua iniziale",
    "Seme",
    "Data di inizio",
    "Data dell'ultima azione",
    colonna_residenza
]

# Individuo le colonne sul cambiamento del patrimonio,
# che appartengono al questionario dei residenti
colonne_patrimonio = [
    colonna
    for colonna in df_turisti_pulito.columns
    if colonna.startswith(
        "A seguito degli investimenti realizzati con il progetto Metroborgo, "
        "ritieni che il valore del tuo patrimonio"
    )
]

# Riunisco tutte le colonne da eliminare
colonne_da_eliminare = colonne_tecniche_turisti + colonne_patrimonio

# Mantengo solamente quelle effettivamente presenti
colonne_da_eliminare = [
    colonna
    for colonna in colonne_da_eliminare
    if colonna in df_turisti_pulito.columns
]

# Elimino le colonne dalla copia pulita
df_turisti_pulito = df_turisti_pulito.drop(
    columns=colonne_da_eliminare
).copy()

print("Colonne eliminate:", len(colonne_da_eliminare))
print("Dimensioni del DataFrame pulito:", df_turisti_pulito.shape)
```

### Verifica del livello di completamento dei questionari

Per ogni questionario viene calcolato il numero di risposte effettivamente compilate.
L’identificativo della risposta viene escluso dal conteggio, perché non rappresenta una domanda del questionario.

Questo controllo permette di individuare i questionari completi, parziali o quasi vuoti prima di decidere se eliminare qualche riga.


```python
# Seleziono solamente le colonne contenenti le domande
colonne_analisi_turisti = [
    colonna
    for colonna in df_turisti_pulito.columns
    if colonna != "ID risposta"
]

# Calcolo, per ogni questionario, quante risposte sono state compilate
completamento_turisti = pd.DataFrame({
    "ID risposta": df_turisti_pulito["ID risposta"],
    "Risposte compilate": df_turisti_pulito[
        colonne_analisi_turisti
    ].notna().sum(axis=1)
})

# Calcolo anche il numero di risposte mancanti
completamento_turisti["Risposte mancanti"] = (
    len(colonne_analisi_turisti)
    - completamento_turisti["Risposte compilate"]
)

# Ordino dai questionari meno completi ai più completi
completamento_turisti = completamento_turisti.sort_values(
    by="Risposte compilate"
)

completamento_turisti
```

### Rimozione dei questionari privi di risposte

Sono stati individuati 11 questionari che non contengono alcuna risposta alle domande utili per l’analisi dei turisti.

Questi questionari vengono esclusi, mentre le compilazioni parziali vengono conservate per una successiva valutazione.


```python
# Individuo le righe senza alcuna risposta alle domande analitiche
indici_questionari_vuoti = completamento_turisti[
    completamento_turisti["Risposte compilate"] == 0
].index

print("Questionari completamente vuoti:",
      len(indici_questionari_vuoti))

# Elimino solamente i questionari senza risposte
df_turisti_pulito = (
    df_turisti_pulito
    .drop(index=indici_questionari_vuoti)
    .reset_index(drop=True)
)

print("Dimensioni aggiornate:",
      df_turisti_pulito.shape)
```

```python
# Individuo i questionari che avevano solamente 5 risposte compilate
id_questionari_parziali = [120, 199]

# Seleziono i due questionari
questionari_parziali = df_turisti_pulito[
    df_turisti_pulito["ID risposta"].isin(id_questionari_parziali)
]

# Per ogni questionario mostro solamente le colonne compilate
for _, riga in questionari_parziali.iterrows():
    print("\n" + "=" * 70)
    print("ID risposta:", riga["ID risposta"])
    print("=" * 70)

    risposte_presenti = riga.dropna()

    for colonna, risposta in risposte_presenti.items():
        if colonna != "ID risposta":
            print(f"\n{colonna}")
            print("Risposta:", risposta)
```

## Profilo sociodemografico dei visitatori

In questa sezione viene descritta la composizione del campione turistico attraverso quattro variabili:

* genere;
* fascia d’età;
* titolo di studio;
* professione.

Per ogni variabile vengono calcolati il numero di rispondenti e la percentuale sul totale delle risposte valide.


```python
# Variabili sociodemografiche da analizzare
variabili_profilo = [
    "Genere",
    "Età:",
    "Titolo di studio",
    "Professione"
]

# Funzione per creare una tabella con conteggi e percentuali
def tabella_percentuali(dataframe, colonna):
    tabella = (
        dataframe[colonna]
        .dropna()
        .value_counts()
        .rename_axis(colonna)
        .reset_index(name="Numero di risposte")
    )

    tabella["Percentuale"] = (
        tabella["Numero di risposte"]
        / tabella["Numero di risposte"].sum()
        * 100
    ).round(1)

    return tabella

# Creo e mostro una tabella per ogni variabile
for variabile in variabili_profilo:
    print("\n" + "=" * 70)
    print(variabile.upper())
    print("=" * 70)

    display(
        tabella_percentuali(
            df_turisti_pulito,
            variabile
        )
    )
```

## Valutazione dell’esperienza turistica

In questa sezione vengono analizzate le affermazioni relative alla qualità e agli effetti percepiti dell’esperienza di visita.

Le risposte sono espresse su una scala da 1 a 7. Per ogni affermazione vengono calcolati:

* il numero di risposte valide;
* il punteggio medio;
* la mediana;
* la percentuale di valutazioni positive, corrispondenti ai punteggi compresi tra 5 e 7.

Valori medi più elevati indicano una percezione maggiormente positiva dell’esperienza.


```python
# Individuo le colonne contenenti le valutazioni su scala da 1 a 7
colonne_valutazioni = [
    colonna
    for colonna in df_turisti_pulito.columns
    if colonna.startswith(
        "Quanto sei d’accordo con le seguenti affermazioni?"
    )
]

# Creo una tabella riepilogativa
riepilogo_valutazioni = []

for colonna in colonne_valutazioni:
    risposte = pd.to_numeric(
        df_turisti_pulito[colonna],
        errors="coerce"
    ).dropna()

    # Estraggo solamente il testo dell'affermazione tra parentesi quadre
    affermazione = colonna.split("[", 1)[-1].rstrip("]")

    riepilogo_valutazioni.append({
        "Affermazione": affermazione,
        "Risposte valide": len(risposte),
        "Media": risposte.mean(),
        "Mediana": risposte.median(),
        "Valutazioni positive (%)": (
            risposte.ge(5).mean() * 100
        )
    })

riepilogo_valutazioni = pd.DataFrame(
    riepilogo_valutazioni
)

# Arrotondo i risultati
riepilogo_valutazioni["Media"] = (
    riepilogo_valutazioni["Media"].round(2)
)

riepilogo_valutazioni["Mediana"] = (
    riepilogo_valutazioni["Mediana"].round(1)
)

riepilogo_valutazioni["Valutazioni positive (%)"] = (
    riepilogo_valutazioni[
        "Valutazioni positive (%)"
    ].round(1)
)

# Ordino dalla valutazione media più alta alla più bassa
riepilogo_valutazioni = (
    riepilogo_valutazioni
    .sort_values("Media", ascending=False)
    .reset_index(drop=True)
)

riepilogo_valutazioni
```

### Sintesi delle valutazioni per area tematica

Le singole affermazioni vengono raggruppate in dimensioni tematiche, così da ottenere una lettura più sintetica dell’esperienza turistica.

Le aree considerate riguardano la qualità e l’accoglienza, il coinvolgimento culturale, la responsabilità verso il territorio, la relazione con il luogo e il benessere percepito.

Per ogni area viene calcolato il punteggio medio ottenuto dalle affermazioni che la compongono.


```python
# Associo a ogni area le relative affermazioni
aree_valutazione = {
    "Qualità e accoglienza": [
        "L’esperienza complessiva è stata di alta qualità",
        "L'esperienza di visita mi ha fatto sentire accolto/a",
        "Gli spazi sono stati chiari e accessibili",
        "L’esperienza è stata autentica"
    ],

    "Apprendimento e coinvolgimento culturale": [
        "Le attività mi hanno aiutato a comprendere il territorio",
        "Ho imparato qualcosa di nuovo",
        "L’esperienza culturale mi ha coinvolto/a, non solo come spettatore/trice",
        "Le storie e i luoghi mi hanno colpito emotivamente",
        "L’esperienza mi ha fatto riflettere"
    ],

    "Responsabilità e sostenibilità": [
        "Ho rispettato i luoghi e le persone",
        "Sono stato/a attento/a all’impatto delle mie azioni",
        "L’esperienza mi ha reso più consapevole del valore del territorio",
        "Mi sentirei responsabile nel tornare come visitatore",
        "Sento di aver contribuito positivamente al territorio"
    ],

    "Relazione con il territorio": [
        "Ho interagito con persone del luogo",
        "Ho utilizzato servizi o prodotti locali",
        "Ho vissuto un senso di connessione con il luogo"
    ],

    "Benessere e intenzione di ritorno": [
        "Consiglierei questa esperienza ad altri",
        "Tornerei o rimarrei più a lungo",
        "L’esperienza mi ha fatto stare bene",
        "Mi sono sentito/a rilassato/a e stimolato/a",
        "Questa esperienza ha arricchito il mio tempo",
        "Me ne vado con una sensazione positiva"
    ]
}

# Uso la tabella già calcolata per ricavare la media di ogni area
riepilogo_aree = []

for area, affermazioni in aree_valutazione.items():
    dati_area = riepilogo_valutazioni[
        riepilogo_valutazioni["Affermazione"].isin(affermazioni)
    ]

    riepilogo_aree.append({
        "Area tematica": area,
        "Numero di affermazioni": len(dati_area),
        "Punteggio medio": dati_area["Media"].mean(),
        "Valutazioni positive (%)":
            dati_area["Valutazioni positive (%)"].mean()
    })

riepilogo_aree = pd.DataFrame(riepilogo_aree)

riepilogo_aree["Punteggio medio"] = (
    riepilogo_aree["Punteggio medio"].round(2)
)

riepilogo_aree["Valutazioni positive (%)"] = (
    riepilogo_aree["Valutazioni positive (%)"].round(1)
)

riepilogo_aree = riepilogo_aree.sort_values(
    "Punteggio medio",
    ascending=False
).reset_index(drop=True)

riepilogo_aree
```

### Distribuzione dei visitatori per fascia di spesa

La spesa dichiarata viene rappresentata considerando tutte le fasce previste dal questionario. Anche le categorie che non hanno ricevuto risposte vengono mantenute nella tabella, così da facilitare il confronto con le rilevazioni future.


```python
# Ordine completo delle fasce di spesa
fasce_spesa = [
    "Meno di 100 €",
    "Tra 100 € e 250 €",
    "Tra 250 € e 500 €",
    "Tra 500 € e 1.000 €",
    "Oltre 1.000 €"
]

# Conteggio delle risposte, includendo anche le fasce con valore zero
conteggio_spesa = (
    df_turisti_pulito[colonna_spesa]
    .value_counts()
    .reindex(fasce_spesa, fill_value=0)
)

# Creo la tabella riepilogativa
tabella_spesa = conteggio_spesa.reset_index()
tabella_spesa.columns = [
    "Fascia di spesa",
    "Numero di risposte"
]

# Calcolo le percentuali sul totale delle risposte valide
tabella_spesa["Percentuale"] = (
    tabella_spesa["Numero di risposte"]
    / tabella_spesa["Numero di risposte"].sum()
    * 100
).round(1)

tabella_spesa
```

### Controllo delle risposte con spesa superiore a 1.000 euro

### Ricostruzione della variabile sul valore patrimoniale

La domanda sul cambiamento del valore patrimoniale è stata esportata in cinque colonne binarie.

La nuova variabile viene ricostruita mantenendo:

* la categoria selezionata, quando è presente una sola risposta;
* un valore mancante, quando non è stata selezionata alcuna opzione;
* una categoria specifica per le due compilazioni che indicano contemporaneamente “È rimasto invariato” ed “È aumentato leggermente”.

In questo modo le risposte ambigue vengono conservate senza attribuire arbitrariamente una categoria.


```python
# Funzione per ricostruire la risposta patrimoniale
def ricostruisci_patrimonio(riga):
    opzioni_selezionate = [
        colonna.split("[", 1)[-1].rstrip("]")
        for colonna in colonne_patrimonio_residenti
        if riga[colonna] == "Sì"
    ]

    # Nessuna risposta selezionata
    if len(opzioni_selezionate) == 0:
        return pd.NA

    # Una sola risposta selezionata
    if len(opzioni_selezionate) == 1:
        return opzioni_selezionate[0]

    # Più risposte selezionate
    return "Risposta multipla: " + " / ".join(opzioni_selezionate)


# Creo la nuova variabile categoriale
df_residenti_pulito["Cambiamento del valore patrimoniale"] = (
    df_residenti_pulito.apply(
        ricostruisci_patrimonio,
        axis=1
    )
)

# Controllo il risultato
df_residenti_pulito[
    "Cambiamento del valore patrimoniale"
].value_counts(dropna=False)
```

In questa sezione vengono analizzate le affermazioni relative alla partecipazione, alla fiducia, al senso di comunità, al legame con il territorio, alla cultura, alla sostenibilità e al benessere percepito.

Le risposte sono espresse su una scala da 1 a 7.

Per ciascuna affermazione vengono calcolati:

* il numero di risposte valide;
* il punteggio medio;
* la mediana;
* la percentuale di valutazioni positive, corrispondenti ai punteggi compresi tra 5 e 7.


```python
# Individuo le colonne con le affermazioni su scala da 1 a 7
colonne_valutazioni_residenti = [
    colonna
    for colonna in df_residenti_pulito.columns
    if colonna.startswith(
        "Quanto sei d’accordo con le seguenti affermazioni?"
    )
]

# Creo il riepilogo delle valutazioni
riepilogo_valutazioni_residenti = []

for colonna in colonne_valutazioni_residenti:
    risposte = pd.to_numeric(
        df_residenti_pulito[colonna],
        errors="coerce"
    ).dropna()

    # Estraggo il testo dell'affermazione
    affermazione = colonna.split("[", 1)[-1].rstrip("]")

    riepilogo_valutazioni_residenti.append({
        "Affermazione": affermazione,
        "Risposte valide": len(risposte),
        "Media": risposte.mean(),
        "Mediana": risposte.median(),
        "Valutazioni positive (%)": (
            risposte.ge(5).mean() * 100
        )
    })

# Trasformo i risultati in una tabella
riepilogo_valutazioni_residenti = pd.DataFrame(
    riepilogo_valutazioni_residenti
)

# Arrotondo i valori
riepilogo_valutazioni_residenti["Media"] = (
    riepilogo_valutazioni_residenti["Media"].round(2)
)

riepilogo_valutazioni_residenti["Mediana"] = (
    riepilogo_valutazioni_residenti["Mediana"].round(1)
)

riepilogo_valutazioni_residenti[
    "Valutazioni positive (%)"
] = (
    riepilogo_valutazioni_residenti[
        "Valutazioni positive (%)"
    ].round(1)
)

# Ordino dalla media più alta alla più bassa
riepilogo_valutazioni_residenti = (
    riepilogo_valutazioni_residenti
    .sort_values("Media", ascending=False)
    .reset_index(drop=True)
)

riepilogo_valutazioni_residenti
```

```python
import math

# Associo il testo breve dell'affermazione
# al nome completo della relativa colonna
mappa_colonne_residenti = {
    colonna.split("[", 1)[-1].rstrip("]"): colonna
    for colonna in colonne_valutazioni_residenti
}

# Raggruppo le affermazioni nelle aree tematiche
aree_residenti = {
    "Iniziativa e partecipazione": [
        "Sento di poter contribuire attivamente alla vita del mio territorio",
        "Quando vedo un problema o un’opportunità nel paese, penso di poter fare qualcosa",
        "Mi sento ascoltato/a quando esprimo idee o proposte sul futuro del territorio",
        "Ho fiducia nella mia capacità di avviare o partecipare a iniziative locali"
    ],

    "Coesione sociale e comunità": [
        "Nel mio territorio le persone collaborano tra loro",
        "Mi sento parte di una comunità",
        "C’è fiducia reciproca tra abitanti, associazioni e operatori locali"
    ],

    "Legame con il territorio": [
        "Mi sento emotivamente legato/a a questo luogo",
        "Sono orgoglioso/a di vivere qui",
        "Questo territorio rappresenta una parte importante di chi sono",
        "Mi dispiacerebbe molto dover lasciare questo luogo"
    ],

    "Cultura e coinvolgimento": [
        "Partecipo ad attività culturali locali (eventi, laboratori, incontri)",
        "Gli spazi culturali mi stimolano a imparare cose nuove",
        "Le attività culturali mi fanno riflettere sul presente e sul futuro del territorio",
        "Mi sento coinvolto/a, non solo spettatore/trice",
        "La cultura ha un ruolo importante nella mia vita quotidiana"
    ],

    "Responsabilità ambientale": [
        "Presto attenzione alla cura degli spazi pubblici",
        "Penso alle conseguenze future delle scelte fatte oggi sul territorio",
        "Nella mia vita quotidiana presto attenzione a ridurre l’impatto ambientale delle mie azioni nel territorio in cui vivo",
        "Cerco di adottare comportamenti rispettosi dell’ambiente negli spazi pubblici e culturali (rifiuti, energia, mobilità, uso delle risorse)"
    ],

    "Benessere e prospettive future": [
        "Mi sento soddisfatto/a della mia vita nel territorio",
        "Vivo questo luogo come un ambiente che favorisce il benessere",
        "Le relazioni sociali contribuiscono positivamente alla mia qualità della vita",
        "Vedo prospettive positive per il futuro della comunità",
        "Mi sento mentalmente ed emotivamente a mio agio qui"
    ]
}

# Creo un DataFrame che conterrà il punteggio
# ottenuto da ogni partecipante nelle diverse aree
punteggi_aree_residenti = pd.DataFrame(
    index=df_residenti_pulito.index
)

for area, affermazioni in aree_residenti.items():

    # Recupero i nomi completi delle colonne
    colonne_area = [
        mappa_colonne_residenti[affermazione]
        for affermazione in affermazioni
    ]

    # Converto le risposte in valori numerici
    valori_area = df_residenti_pulito[
        colonne_area
    ].apply(
        pd.to_numeric,
        errors="coerce"
    )

    # Richiedo almeno metà delle affermazioni compilate
    minimo_risposte = math.ceil(len(colonne_area) / 2)

    # Calcolo la media individuale dell'area
    punteggio_area = valori_area.mean(axis=1)

    # Imposto come mancante il punteggio di chi
    # non ha risposto ad almeno metà delle affermazioni
    punteggio_area[
        valori_area.notna().sum(axis=1) < minimo_risposte
    ] = pd.NA

    punteggi_aree_residenti[area] = punteggio_area


# Creo il riepilogo generale delle aree
riepilogo_aree_residenti = []

for area in punteggi_aree_residenti.columns:
    punteggi_validi = pd.to_numeric(
        punteggi_aree_residenti[area],
        errors="coerce"
    ).dropna()

    riepilogo_aree_residenti.append({
        "Area tematica": area,
        "Numero di affermazioni": len(aree_residenti[area]),
        "Risposte valide": len(punteggi_validi),
        "Punteggio medio": punteggi_validi.mean(),
        "Mediana": punteggi_validi.median(),
        "Valutazioni positive (%)": (
            punteggi_validi.ge(5).mean() * 100
        )
    })

riepilogo_aree_residenti = pd.DataFrame(
    riepilogo_aree_residenti
)

# Arrotondo i risultati
riepilogo_aree_residenti["Punteggio medio"] = (
    riepilogo_aree_residenti["Punteggio medio"].round(2)
)

riepilogo_aree_residenti["Mediana"] = (
    riepilogo_aree_residenti["Mediana"].round(2)
)

riepilogo_aree_residenti["Valutazioni positive (%)"] = (
    riepilogo_aree_residenti[
        "Valutazioni positive (%)"
    ].round(1)
)

# Ordino dalla valutazione più alta alla più bassa
riepilogo_aree_residenti = (
    riepilogo_aree_residenti
    .sort_values(
        "Punteggio medio",
        ascending=False
    )
    .reset_index(drop=True)
)

riepilogo_aree_residenti
```
